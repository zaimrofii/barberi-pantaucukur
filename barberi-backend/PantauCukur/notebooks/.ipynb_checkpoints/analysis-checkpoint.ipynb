```python
import json
import pandas as pd

LOG_PATH = r'C:\Users\zaimr\projects\barberi-pantaucukur\barberi-backend\PantauCukur\core\logs\ai_engine.jsonl'

records = []
errors = 0

with open(LOG_PATH, 'r', encoding='utf-8') as f:
    for i, line in enumerate(f, 1):
        line = line.strip()
        if not line:
            continue
        try:
            records.append(json.loads(line))
        except json.JSONDecodeError:
            errors += 1
            print(f"⚠️ Baris {i} corrupt, di-skip")

df = pd.DataFrame(records)
print(f"✅ Total baris valid: {len(df)} | Corrupt: {errors}")
df.head()
```

```python
import pandas as pd
import json

pd.set_option('display.max_colwidth', 500)  # atau None untuk unlimited
pd.set_option('display.max_rows', 500)      # atau None untuk unlimited

alamat = r'C:\Users\zaimr\projects\barberi-pantaucukur\barberi-backend\PantauCukur\core\logs\ai_engine.jsonl'
data = []

with open(alamat, 'r', encoding='utf-8') as f :
    for line in f:
        row = json.loads(line)
        data.append(row)

df = pd.DataFrame(data)
df.head()
```

```python
print(df['text'].head())
```

```python
print(df['record'][0])
```

```python
df.info()
```

```python
df['record'].info()
```

```python
df['record'].describe()
```

```python
# Ambil text dan record
df['raw_text'] = df['text']  # teks asli
df['record'] = df['record']  # metadata

# Ekstrak dari record
df['timestamp'] = df['record'].apply(lambda x: x['time']['timestamp'])
df['level'] = df['record'].apply(lambda x: x['level']['name'])
df['file_name'] = df['record'].apply(lambda x: x['file']['name'])
df['function'] = df['record'].apply(lambda x: x['function'])
df['elapsed_seconds'] = df['record'].apply(lambda x: x['elapsed']['seconds'])

print(df)
```
